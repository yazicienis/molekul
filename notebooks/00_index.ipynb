## Environment check

Run this cell first. It confirms `molekul` imports, reports which optional packages
you have, and runs a 1-second smoke test (the RHF energy of water).

````python
# --- Make the MOLEKUL package importable (same bootstrap every notebook uses) ---
try:
    import molekul  # noqa: F401
except ModuleNotFoundError:
    import sys, pathlib
    for _p in (pathlib.Path.cwd(), *pathlib.Path.cwd().parents):
        if (_p / "src" / "molekul").is_dir():
            sys.path.insert(0, str(_p / "src"))
            break
    import molekul  # noqa: F401

import importlib.util as _u
import platform

print("Python      :", platform.python_version())
print("molekul     : OK   (core needs only numpy + scipy)")
for _mod, _what in [
    ("matplotlib", "plots - used by every notebook"),
    ("pyscf",      "optional - 'compare to a reference code' cells"),
    ("py3Dmol",    "optional - inline 3D viewer (notebook 06)"),
]:
    _ok = _u.find_spec(_mod) is not None
    print(f"{_mod:11s} :", "OK  " if _ok else "----", "-", _what)

# --- 1-second smoke test: RHF/STO-3G water should be about -74.9630 Ha ---
import numpy as np
from molekul.molecule import Molecule
from molekul.atoms import Atom
from molekul.basis_sto3g import STO3G
from molekul.rhf import rhf_scf

_h2o = Molecule(atoms=[
    Atom.from_angstrom("O",  0.0000, 0.0, 0.1173),
    Atom.from_angstrom("H",  0.7572, 0.0, -0.4692),
    Atom.from_angstrom("H", -0.7572, 0.0, -0.4692),
])
_e = rhf_scf(_h2o, STO3G).energy_total
print(f"\nSmoke test  : RHF/STO-3G water = {_e:.6f} Ha  (expected ~ -74.9630)")
assert abs(_e - (-74.963023)) < 1e-3, "Unexpected energy - check your installation."
print("Environment looks good. Open 01_atoms_molecules_constants.ipynb next.")
````
